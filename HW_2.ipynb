```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix

# Load MNIST
data = fetch_openml('mnist_784', version=1, as_frame=False)
X = data.data
y = data.target.astype(int)

# Keep only even-numbered digits
even = y % 2 == 0
X = X[even]
y = y[even]

# X has 784 columns, so every pixel is used
print('X shape:', X.shape)

# Split the data
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=123,
    shuffle=True,
    stratify=y
)

# Train the k-nearest neighbors classifier
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_train, y_train)

# Predict the test data
preds = model.predict(X_test)

print('Accuracy score:', accuracy_score(y_test, preds))

# Confusion-matrix heatmap
even_digits = [0, 2, 4, 6, 8]
cm = confusion_matrix(y_test, preds, labels=even_digits)

ax = sns.heatmap(
    cm,
    linewidths=2,
    annot=True,
    fmt='d',
    cmap='viridis',
    cbar=True
)

ax.set_xticklabels(even_digits)
ax.set_yticklabels(even_digits)
plt.xlabel('Predicted Digit')
plt.ylabel('Actual Digit')
plt.title('kNN Classification Results')
plt.show()

# Correctly classified 6's out of the total number of 6's
correct_sixes = cm[even_digits.index(6), even_digits.index(6)]
total_sixes = np.sum(y_test == 6)

print("Correctly classified 6's:", correct_sixes, "out of", total_sixes)
```

```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans as SKKMeans
from sklearn.metrics import silhouette_score, adjusted_rand_score
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# Prepare the data


np.random.seed(123)

data = load_iris()

df = pd.DataFrame()
df['sepal length'] = data['data'][:, 0]
df['petal length'] = data['data'][:, 2]
df['petal width'] = data['data'][:, 3]

X = df.to_numpy()       # three features used by k-means
y = data['target']      # only used to compare with the known species
df['species_name'] = [data.target_names[i] for i in y]

print(df.head())
print('\nX shape:', X.shape)


# 3D plot of the actual Iris species


fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

for i, species in enumerate(data.target_names):
    points = X[y == i]
    ax.scatter(points[:, 0], points[:, 1], points[:, 2], label=species)

ax.set_xlabel('sepal length')
ax.set_ylabel('petal length')
ax.set_zlabel('petal width')
ax.set_title('Iris data: actual species')
ax.legend()
plt.show()


# Our k-means algorithm


class KMeans:
    def __init__(self, k):
        self.k = k
        self.cluster_labels = None
        self.centroids = None
        self.clusters = None

    def fit(self, X):
        self.centroids = X[np.random.choice(X.shape[0], self.k, replace=False), :]
        self.cluster_labels = np.arange(self.k)

        while True:
            distances = np.array([
                np.linalg.norm(X - centroid, axis=1)
                for centroid in self.centroids
            ])

            self.clusters = np.argmin(distances, axis=0)

            new_centroids = np.array([
                X[self.clusters == i, :].mean(axis=0)
                for i in range(self.k)
            ])

            if np.allclose(new_centroids, self.centroids):
                break
            else:
                self.centroids = new_centroids

    def predict(self, X):
        distances = np.array([
            np.linalg.norm(X - centroid, axis=1)
            for centroid in self.centroids
        ])

        return self.cluster_labels[np.argmin(distances, axis=0)]

    def calculate_inertia(self, X):
        if self.centroids is None or self.clusters is None:
            raise ValueError("Model must be fitted before calculating inertia.")
        inertia = 0
        for i in range(self.k):
            cluster_points = X[self.clusters == i]
            if len(cluster_points) > 0:
                inertia += np.sum(np.linalg.norm(cluster_points - self.centroids[i], axis=1)**2)
        return inertia


# Fit our model using three clusters
model = KMeans(k=3)
model.fit(X)
our_labels = model.predict(X)


# Fit sklearn's k-means model


skmodel = SKKMeans(
    n_clusters=3,
    init='k-means++',
    n_init=10,
    random_state=1234
)

skmodel.fit(X)
sklearn_labels = skmodel.predict(X)


# Compare our method with sklearn using 3D plots


fig = plt.figure(figsize=(15, 6))

ax1 = fig.add_subplot(121, projection='3d')
ax1.scatter(X[:, 0], X[:, 1], X[:, 2], c=our_labels, cmap='viridis')
ax1.scatter(
    model.centroids[:, 0],
    model.centroids[:, 1],
    model.centroids[:, 2],
    color='red', marker='x', s=150, linewidths=3
)
ax1.set_xlabel('sepal length')
ax1.set_ylabel('petal length')
ax1.set_zlabel('petal width')
ax1.set_title('Our k-means, k = 3')

ax2 = fig.add_subplot(122, projection='3d')
ax2.scatter(X[:, 0], X[:, 1], X[:, 2], c=sklearn_labels, cmap='viridis')
ax2.scatter(
    skmodel.cluster_centers_[:, 0],
    skmodel.cluster_centers_[:, 1],
    skmodel.cluster_centers_[:, 2],
    color='red', marker='x', s=150, linewidths=3
)
ax2.set_xlabel('sepal length')
ax2.set_ylabel('petal length')
ax2.set_zlabel('petal width')
ax2.set_title('sklearn k-means, k = 3')

plt.tight_layout()
plt.show()


# Cluster numbers may be different even when the groupings are the same.
# These scores do not depend on the names of the cluster numbers.
comparison = pd.DataFrame({
    'method': ['Our k-means', 'sklearn k-means'],
    'silhouette score': [
        silhouette_score(X, our_labels),
        silhouette_score(X, sklearn_labels)
    ],
    'adjusted Rand score': [
        adjusted_rand_score(y, our_labels),
        adjusted_rand_score(y, sklearn_labels)
    ]
})

print('\nPerformance comparison:')
print(comparison.round(3).to_string(index=False))


# Vary the number of clusters: k = 2, 3, and 4


fig = plt.figure(figsize=(18, 5))

for plot_number, k in enumerate([2, 3, 4], start=1):
    np.random.seed(123)
    test_model = KMeans(k=k)
    test_model.fit(X)
    labels = test_model.predict(X)

    ax = fig.add_subplot(1, 3, plot_number, projection='3d')
    ax.scatter(X[:, 0], X[:, 1], X[:, 2], c=labels, cmap='viridis')
    ax.scatter(
        test_model.centroids[:, 0],
        test_model.centroids[:, 1],
        test_model.centroids[:, 2],
        color='red', marker='x', s=130, linewidths=3
    )
    ax.set_xlabel('sepal length')
    ax.set_ylabel('petal length')
    ax.set_zlabel('petal width')
    ax.set_title('Our k-means, k = ' + str(k))

plt.tight_layout()
plt.show()


# Elbow plot and silhouette plot for different values of k


distortions = []
silhouette_scores = []

for k in range(2, 9):
    kmeans = SKKMeans(
        n_clusters=k,
        init='k-means++',
        n_init=10,
        random_state=123
    )
    kmeans.fit(X)
    distortions.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(X, kmeans.labels_))

our_distortions = []
our_silhouette_scores = []

for k in range(2, 9):
    # Using a fixed seed for our custom KMeans for reproducibility
    np.random.seed(123)
    our_kmeans = KMeans(k=k)
    our_kmeans.fit(X)
    our_distortions.append(our_kmeans.calculate_inertia(X))
    our_silhouette_scores.append(silhouette_score(X, our_kmeans.clusters))


fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(range(2, 9), distortions, marker='o', label='sklearn k-means')
axes[0].plot(range(2, 9), our_distortions, marker='x', label='Our k-means')
axes[0].grid(True)
axes[0].set_xticks(range(2, 9))
axes[0].set_title('Elbow plot')
axes[0].set_xlabel('number of clusters, k')
axes[0].set_ylabel('inertia')
axes[0].legend() # Add legend

axes[1].plot(range(2, 9), silhouette_scores, marker='o', label='sklearn k-means')
axes[1].plot(range(2, 9), our_silhouette_scores, marker='x', label='Our k-means')
axes[1].grid(True)
axes[1].set_xticks(range(2, 9))
axes[1].set_title('Silhouette scores')
axes[1].set_xlabel('number of clusters, k')
axes[1].set_ylabel('silhouette score')
axes[1].legend() # Add legend

plt.tight_layout()
plt.show()
```
